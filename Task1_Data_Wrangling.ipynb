```python

# TASK 1: Data Immersion & Wrangling
# Dataset: sales_data.csv


import pandas as pd
from google.colab import files

uploaded = files.upload()

df = pd.read_csv("sales_data.csv")

print("First 5 rows of dataset:")
display(df.head())

print("\nDataset Info:")
df.info()

print("\nSummary Statistics:")
display(df.describe())

print("\nMissing Values Check:")
display(df.isnull().sum())

print("\nDuplicate Rows Check:")
print("Number of duplicate rows:", df.duplicated().sum())


#Remove Duplicates
df = df.drop_duplicates()

print("\nShape After Removing Duplicates:", df.shape)


# Data Validation Checks
print("\nNegative Value Check:")
display((df[['total_bill','tip']] < 0).sum())

print("\nInvalid Tip Check (tip > total_bill):")
print((df['tip'] > df['total_bill']).sum())


# Tip percentage (Decimal format for Data Science)
df['tip_percentage'] = df['tip'] / df['total_bill']

df['revenue_per_person'] = df['total_bill'] / df['size']
df['tip_per_person'] = df['tip'] / df['size']

print("\nFeature Engineering Completed.")
display(df.head())


#Data Type Optimization
df['sex'] = df['sex'].astype('category')
df['smoker'] = df['smoker'].astype('category')
df['day'] = df['day'].astype('category')
df['time'] = df['time'].astype('category')

print("\nUpdated Data Types:")
df.info()


print("\nFinal Dataset Shape:", df.shape)
print("\nFinal Dataset Preview:")
display(df.head())


df.to_csv("cleaned_sales_data.csv", index=False)

print("\nCleaned dataset exported successfully.")

files.download("cleaned_sales_data.csv")
```
